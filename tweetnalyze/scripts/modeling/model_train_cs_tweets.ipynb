```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('../../data/carbon_sustain_tweets_impression.sqlite3')

df = pd.read_sql_query("SELECT * FROM app_tweet", conn)

print(df.head())

conn.close()
```

```python
# Assuming df is already loaded
df = df[~df['text'].str.startswith("RT")]
```

```python
import numpy as np

df['like_count_log'] = np.log1p(df['like_count'])  # log(1 + x)
df['impression_count_log'] = np.log1p(df['impression_count'])  # log(1 + x)
```

```python
X = df.drop(["id", "tweet_id", "retweet_count", "reply_count", "quote_count", 
             "possibly_sensitive", "user_id", "like_count", "impression_count", "created_at", "url_link_clicks", "user_profile_clicks",
            "bookmark_count", "hashtags", "mentions", "urls", "symbols"], axis=1)
# Y = df["like_count_log"].values
Y = df["impression_count_log"].values
```

```python
import re
import string
import pandas as pd

# Function to clean tweets
def clean_tweet(text):
    text = str(text)  # Just in case of NaN
    text = text.lower()  # Lowercase
    text = re.sub(r'http\S+|www\S+|https\S+', '', text, flags=re.MULTILINE)  # Remove URLs
    text = re.sub(r'@\w+', '', text)  # Remove @mentions
    text = re.sub(r'#', '', text)  # Remove hashtag symbols, keep the word
    text = re.sub(r'RT[\s]+', '', text)  # Remove "RT"
    text = re.sub(r'\n', ' ', text)  # Remove line breaks
    text = re.sub(r'[^A-Za-z0-9\s]', '', text)  # Remove emojis and special characters
    text = re.sub(r'\s+', ' ', text).strip()  # Remove extra spaces
    return text
```

```python
def clean_for_transformers(text):
    text = str(text)
    text = re.sub(r"http\S+|www\S+", "", text)  # remove URLs
    text = re.sub(r"\n", " ", text)  # remove line breaks
    text = re.sub(r'\s+', ' ', text).strip()  # remove extra spaces
    return text
```

```python
X['clean_text'] = X['text'].apply(clean_tweet)
X['clean_text_transformers'] = X['text'].apply(clean_for_transformers)
```

```python
X
```

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer('all-MiniLM-L6-v2')
X_vec = model.encode(X['clean_text_transformers'].tolist(), convert_to_tensor=True).cpu().numpy()
```

```python
print(X_vec.shape)
```

```python
X_vec
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_vec, Y, test_size=0.2, random_state=42)
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

reg = RandomForestRegressor(n_estimators=100, random_state=42)
reg.fit(X_train, y_train)

# Predict
y_pred = reg.predict(X_test)

# Evaluate
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression()
reg.fit(X_train, y_train)
# Predict
y_pred = reg.predict(X_test)

# Evaluate
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

```python
from sklearn.linear_model import Ridge, Lasso
reg = Ridge(alpha=1.0)  # L2 regularization
reg.fit(X_train, y_train)
# Predict
y_pred = reg.predict(X_test)

# Evaluate
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

```python
from xgboost import XGBRegressor
reg = XGBRegressor()
reg.fit(X_train, y_train)
# Predict
y_pred = reg.predict(X_test)

# Evaluate
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```

```python
from lightgbm import LGBMRegressor
reg = LGBMRegressor()
reg.fit(X_train, y_train)
# Predict
y_pred = reg.predict(X_test)

# Evaluate
print("MSE:", mean_squared_error(y_test, y_pred))
print("R² Score:", r2_score(y_test, y_pred))
```
